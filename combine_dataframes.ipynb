```python
import pandas as pd
import re

name_model = "llama-3.3-70b-instruct"

paths = [
    f"C:/Users/dnnxl/Documents/GitHub/TrackEval/DJI_0005_V_Aug/trackers/mot_challenge/MOT15-all/{name_model}/pedestrian_detailed.csv",
    f"C:/Users/dnnxl/Documents/GitHub/TrackEval/DJI_0006_V_2_Aug/trackers/mot_challenge/MOT15-all/{name_model}/pedestrian_detailed.csv",
    f"C:/Users/dnnxl/Documents/GitHub/TrackEval/DJI_0007_V_1_Aug/trackers/mot_challenge/MOT15-all/{name_model}/pedestrian_detailed.csv",
    f"C:/Users/dnnxl/Documents/GitHub/TrackEval/DJI_0010_V_1_Aug/trackers/mot_challenge/MOT15-all/{name_model}/pedestrian_detailed.csv",
    f"C:/Users/dnnxl/Documents/GitHub/TrackEval/DJI_0010_V_2_Aug/trackers/mot_challenge/MOT15-all/{name_model}/pedestrian_detailed.csv"
]

pattern = r"DJI_\d+_V(?:_\d+)?_Aug"
videonames = [re.search(pattern, path).group() for path in paths if re.search(pattern, path)]

list_dfs = []
for file, video_name in zip(paths, videonames):
    df = pd.read_csv(file)
    df["Training"] = ["vertical"] * df.shape[0]
    df["Video_Training"] = [video_name] * df.shape[0]
    list_dfs.append(df)

df_final = pd.concat(list_dfs, ignore_index=True)
df_final = df_final[df_final["seq"] != "COMBINED"]
df_final.to_csv(f"{name_model}.csv")
```
